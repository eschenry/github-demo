A. Write a Python program to find the factorial of a number using a while loop.

Example guide on how to compute for a factorial:

n! = n × (n−1) × (n−2) × ... × 1

So for 4!

4!=4×3×2×1 = 24

```python
print("Here is a program to calculate the factorial of a number:")

# Get input
num = int(input("Enter a number: "))

# Initialize variables
orignum = num
factorial = num
i = num

# Check if the number is negative
if num < 0:
    print(f"Factorial does not exist for negative number, {num}.")
elif num == 0:
    print("The factorial of 0 is 1.")
else:
    while i != 1:
        factorial = num * (i - 1)
        # print(f"Computing this step, {num} * ({i - 1}) is {factorial}.") #just to check the step, can be commented out
        num = factorial
        i -= 1

    print()
    print(f"The factorial of {orignum} is {factorial}.")
```
